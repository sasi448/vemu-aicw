```python
#step1 : importing the libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#step5 visualization of data set
import matplotlib.pyplot as plt
plt.figure(figsize=(5,6))
plt.scatter(r_df['age'],r_df['balance'],color='red',s=60)
plt.show()
#interpretation
# most of the balance is between 0 -20000 and highest balance is nearly 80000 for the age range fo 80-90 
# for 50-60 the highest balance is approximately 60000
# nexxt the balance ranges moderatly in range 40000-60000 balance
```

```python
#step6 : define target and features
x = r_df[['age']]
y = r_df[['balance']]
```

```python
#step 7 : splitting dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2 , random_state = 42)
```

```python
#step 8 model creation linear regression
from sklearn.linear_model import LinearRegression
model = LinearRegression()
fitted_model = model.fit(x_train,y_train)
```

```python
#step 9 Identifying model parameters 
print('slope: ',fitted_model.coef_)
print('Intercept : ' , fitted_model.intercept_)
```

```python
#step 10 prediction from test data 
y_pred_lrm = fitted_model.predict(x_test)
```

```python
print(y_pred_lrm)
```

```python
#step 11 characteristics of actual vs predictions
com = pd.DataFrame({
    "Actual": np.array(y_test).ravel(),
    "Predicted": y_pred_lrm.ravel()
})

print(com)
```

```python
#stp 12 model evaluation
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
mae =mean_absolute_error(y_test,y_pred_lrm)
mse = mean_squared_error(y_test,y_pred_lrm)
smse = np.sqrt(mse)
r2  = r2_score(y_test,y_pred_lrm)
print("mean_absolute_error : ",mae)
print("mean_squared_error : ",mse)
print("square root of mean_squared_error : ",smse)
print("r2_score : ",r2)
```
