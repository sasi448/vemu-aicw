```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
rdf=pd.read_csv("C:\\Users\\HP\\Downloads//bsv.csv")
```

```python
# Step 3: Description of data
rdf.head()
rdf.tail()
rdf.info()
rdf.shape
rdf.describe()
```

```python
rdf.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(3,4))
plt.scatter(rdf["age"],rdf["balance"],color="red",s=60)
plt.show()
```

```python
 #Step 6: Define Features and Target
x=rdf[["age"]]
y=rdf["balance"]
```

```python
# Step 7: Splitting dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,
                                               test_size=0.2,
                                               random_state=42)
```

```python
# Step 8: Model creation : Linear regression
from sklearn.linear_model import LinearRegression
lrm=LinearRegression()
fitted_lrm=lrm.fit(x_train,y_train)
```

```python
# Step 9: Identifying Model parameters
print("Slope:",fitted_lrm.coef_)
print("Intercept:",fitted_lrm.intercept_)
```

```python
# Step 10: Predections from test data
y_pre=fitted_lrm.predict(x_test)
print(y_pre)
```

```python
# Step 11 : Comparision of Actual Vs Predections
com=pd.DataFrame({"Actual":y_test,
                "Predected":y_pre})
print(com)
```

```python
# Step 12: Model Evaluation
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error 
from sklearn.metrics import r2_score
import numpy as np

mae=mean_absolute_error(y_test,y_pre)
mse=mean_squared_error (y_test,y_pre)
smse=np.sqrt(mse)
r2=r2_score(y_test,y_pre)
print(mae)
print(mse)
print(smse)
print(r2)
```
