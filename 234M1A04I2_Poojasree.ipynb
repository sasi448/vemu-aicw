```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
rdf=pd.read_csv("C:\\Users\Kedhari\Documents//bank.csv")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(3,4))
plt.scatter(rdf["age"],rdf["balance"],color="blue",s=60)
plt.show()
```

```python
x=rdf[["age"]]
y=rdf[["balance"]]
```

```python
#step7:splitting data set
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#interpretation:80% training data,testing data
```

```python

#step:8 model creation:liner regression
from sklearn.linear_model import LinearRegression
lrm=LinearRegression()
fitted_lrm=lrm.fit(x_train,y_train)
```

```python
#step9:identifying model parameters
print("Slope:",fitted_lrm.coef_)
print("Intercept:",fitted_lrm.intercept_)
```

```python
#step10:predections from test data
y_pre=fitted_lrm.predict(x_test)
```

```python
print(y_pre)
```

```python
#step12:model evaluation
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
mae= mean_absolute_error(y_test,y_pre)
mse=mean_squared_error(y_test,y_pre)
smse=np.sqrt(mse)
r2=r2_score(y_test,y_pre)
print(mae)
print(mse)
print(smse)
print(r2)
```
